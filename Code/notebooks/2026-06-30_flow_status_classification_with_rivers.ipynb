# Reusable JRC + HydroRIVERS workflow

This notebook is organized as a reusable pipeline. Change `config` first, then run the workflow cells below.


```python
from dataclasses import dataclass, replace
from pathlib import Path
import math

import fiona
import geopandas as gpd
import matplotlib.colors as mcolors
import matplotlib.patches as mpatches
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import rasterio
from matplotlib.colors import BoundaryNorm, ListedColormap
from matplotlib.lines import Line2D
from rasterio.features import geometry_mask
from rasterio.merge import merge
from rasterio.transform import Affine, rowcol
from rasterio.windows import from_bounds
from scipy.ndimage import label
from shapely.geometry import box


@dataclass(frozen=True)
class AnalysisConfig:
    season_path: Path = Path("../../Data/seasonality_110E_40Nv1_4_2021.tif")
    hydrorivers_gdb_path: Path = Path("../../Data/Google Earth Engine/HydroRIVERS_v10_as.gdb/HydroRIVERS_v10_as.gdb")
    bounds: tuple[float, float, float, float] = (115, 30, 120, 35)  # lon_min, lat_min, lon_max, lat_max

    # 33 is roughly 1 km for this raster; 333 is roughly 10 km.
    factor: int = 33
    water_ratio_threshold: float = 0.4
    perennial_month_threshold: int = 10
    perennial_ratio_threshold: float = 0.5

    discharge_threshold_cms: float = 40
    buffer_cells: int = 5
    cell_size_m: int = 1000
    endpoint_radius_cells: int = 5
    majority_perennial_threshold: float = 0.7
    fallback_metric_crs: str = "EPSG:32650"

    @property
    def lon_min(self):
        return self.bounds[0]

    @property
    def lat_min(self):
        return self.bounds[1]

    @property
    def lon_max(self):
        return self.bounds[2]

    @property
    def lat_max(self):
        return self.bounds[3]

    @property
    def buffer_m(self):
        return self.buffer_cells * self.cell_size_m


config = AnalysisConfig()
```

```python
def read_seasonality_window(season_path, bounds):
    """Read a JRC seasonality raster window and convert 255 nodata values to NaN."""
    lon_min, lat_min, lon_max, lat_max = bounds

    with rasterio.open(season_path) as src:
        window = from_bounds(lon_min, lat_min, lon_max, lat_max, transform=src.transform)
        window = window.round_offsets().round_lengths()

        season = src.read(1, window=window).astype("float32")
        season[season == 255] = np.nan

        transform = src.window_transform(window)
        crs = src.crs

    return season, transform, crs


def pad_to_factor(array, factor, fill_value=np.nan):
    """Pad a 2D array so its height and width are divisible by factor."""
    height, width = array.shape
    new_height = math.ceil(height / factor) * factor
    new_width = math.ceil(width / factor) * factor

    padded = np.full((new_height, new_width), fill_value, dtype=array.dtype)
    padded[:height, :width] = array
    return padded


def aggregate_water_binary(season, factor, threshold):
    """Aggregate seasonality pixels into water / non-water grid cells."""
    padded = pad_to_factor(season, factor, fill_value=np.nan)
    valid = ~np.isnan(padded)
    water = (padded > 0) & valid

    block_h = padded.shape[0] // factor
    block_w = padded.shape[1] // factor

    valid_blocks = valid.reshape(block_h, factor, block_w, factor)
    water_blocks = water.reshape(block_h, factor, block_w, factor)

    valid_count = valid_blocks.sum(axis=(1, 3))
    water_count = water_blocks.sum(axis=(1, 3))

    water_ratio = np.divide(
        water_count,
        valid_count,
        out=np.zeros_like(water_count, dtype="float32"),
        where=valid_count > 0,
    )
    water_binary = (water_ratio > threshold).astype("uint8")

    return water_binary, water_ratio, valid_count


def aggregate_perennial_status(
    season,
    water_agg,
    factor,
    perennial_threshold,
    ratio_threshold,
):
    """Classify aggregated cells as non-water, intermittent, or perennial."""
    padded = pad_to_factor(season, factor, fill_value=np.nan)
    valid = ~np.isnan(padded)

    water_pixel = (padded > 0) & valid
    perennial_pixel = (padded >= perennial_threshold) & valid

    block_h = padded.shape[0] // factor
    block_w = padded.shape[1] // factor

    water_blocks = water_pixel.reshape(block_h, factor, block_w, factor)
    perennial_blocks = perennial_pixel.reshape(block_h, factor, block_w, factor)

    water_count = water_blocks.sum(axis=(1, 3))
    perennial_count = perennial_blocks.sum(axis=(1, 3))

    perennial_ratio = np.divide(
        perennial_count,
        water_count,
        out=np.zeros_like(perennial_count, dtype="float32"),
        where=water_count > 0,
    )

    status = np.zeros_like(water_agg, dtype="uint8")
    status[water_agg == 1] = 1
    status[(water_agg == 1) & (perennial_ratio > ratio_threshold)] = 2

    return status, perennial_ratio, water_count, perennial_count


def aggregated_transform(source_transform, factor):
    return source_transform * Affine.scale(factor, factor)


def bounds_to_extent(bounds):
    lon_min, lat_min, lon_max, lat_max = bounds
    return [lon_min, lon_max, lat_min, lat_max]


def raster_extent(transform, shape):
    height, width = shape
    west = transform.c
    north = transform.f
    east = west + transform.a * width
    south = north + transform.e * height
    return [west, east, south, north]


def summarize_array(name, array):
    values = np.unique(array[~np.isnan(array)]) if np.issubdtype(array.dtype, np.floating) else np.unique(array)
    print(f"{name} shape: {array.shape}")
    print(f"{name} values: {values}")


def save_flow_status_geotiff(flow_status, transform, crs, output_path, nodata=255):
    """Save flow_status as a georeferenced GeoTIFF for reuse and mosaicking.

    Values:
    0 = non-water
    1 = intermittent
    2 = perennial
    255 = nodata
    """
    output_path = Path(output_path)
    output_path.parent.mkdir(parents=True, exist_ok=True)

    data = flow_status.astype("uint8")
    with rasterio.open(
        output_path,
        "w",
        driver="GTiff",
        height=data.shape[0],
        width=data.shape[1],
        count=1,
        dtype="uint8",
        crs=crs,
        transform=transform,
        nodata=nodata,
        compress="lzw",
    ) as dst:
        dst.write(data, 1)
        dst.update_tags(
            1,
            value_0="non-water",
            value_1="intermittent",
            value_2="perennial",
            value_255="nodata",
        )

    return output_path


def mosaic_flow_status_geotiffs(input_paths, output_path, nodata=255):
    """Merge multiple flow_status GeoTIFF tiles into one georeferenced GeoTIFF."""
    input_paths = [Path(path) for path in input_paths]
    output_path = Path(output_path)
    output_path.parent.mkdir(parents=True, exist_ok=True)

    datasets = [rasterio.open(path) for path in input_paths]
    try:
        mosaic, mosaic_transform = merge(datasets, nodata=nodata)
        profile = datasets[0].profile.copy()
        profile.update(
            height=mosaic.shape[1],
            width=mosaic.shape[2],
            transform=mosaic_transform,
            nodata=nodata,
            compress="lzw",
        )

        with rasterio.open(output_path, "w", **profile) as dst:
            dst.write(mosaic)
    finally:
        for dataset in datasets:
            dataset.close()

    return output_path
```

```python
def flow_status_colormap():
    cmap = ListedColormap(["lightgray", "orange", "blue"])
    norm = BoundaryNorm([-0.5, 0.5, 1.5, 2.5], cmap.N)
    return cmap, norm


def plot_seasonality(season, bounds, ax=None):
    if ax is None:
        fig, ax = plt.subplots(figsize=(8, 8))
    else:
        fig = ax.figure

    season_cmap = plt.get_cmap("viridis", 13)
    season_norm = mcolors.BoundaryNorm(np.arange(-0.5, 13.5, 1), season_cmap.N)

    im = ax.imshow(
        season,
        extent=bounds_to_extent(bounds),
        origin="upper",
        cmap=season_cmap,
        norm=season_norm,
    )
    fig.colorbar(im, ax=ax, label="Water months per year", ticks=np.arange(13))
    ax.set_title("JRC Water Seasonality")
    ax.set_xlabel("Longitude")
    ax.set_ylabel("Latitude")
    return fig, ax


def plot_water_binary(water_agg, extent, factor, ax=None):
    if ax is None:
        fig, ax = plt.subplots(figsize=(8, 8))
    else:
        fig = ax.figure

    cmap = plt.get_cmap("gray_r", 2)
    norm = mcolors.BoundaryNorm([-0.5, 0.5, 1.5], cmap.N)

    ax.imshow(water_agg, extent=extent, origin="upper", cmap=cmap, norm=norm)
    ax.legend(
        handles=[
            mpatches.Patch(color=cmap(0), label="Non-water"),
            mpatches.Patch(color=cmap(1), label="Water"),
        ],
        loc="upper right",
    )
    ax.set_title(f"Aggregated Water / Non-water Map, factor={factor}")
    ax.set_xlabel("Longitude")
    ax.set_ylabel("Latitude")
    return fig, ax


def plot_flow_status(flow_status, extent, factor, perennial_month_threshold, ax=None, alpha=1):
    if ax is None:
        fig, ax = plt.subplots(figsize=(8, 8))
    else:
        fig = ax.figure

    cmap, norm = flow_status_colormap()
    ax.imshow(flow_status, extent=extent, origin="upper", cmap=cmap, norm=norm, alpha=alpha)
    ax.legend(
        handles=[
            mpatches.Patch(color="lightgray", label="Non-water"),
            mpatches.Patch(color="orange", label="Intermittent"),
            mpatches.Patch(color="blue", label="Perennial"),
        ],
        loc="upper right",
        title="Flow status",
    )
    ax.set_title(f"Aggregated Flow Status, factor={factor}, perennial >= {perennial_month_threshold} months")
    ax.set_xlabel("Longitude")
    ax.set_ylabel("Latitude")
    return fig, ax


def plot_flow_with_rivers(flow_status, extent, region_gdf, rivers_region, config, ax=None):
    if ax is None:
        fig, ax = plt.subplots(figsize=(8, 8))
    else:
        fig = ax.figure

    cmap, norm = flow_status_colormap()
    ax.imshow(flow_status, extent=extent, origin="upper", cmap=cmap, norm=norm)
    region_gdf.boundary.plot(ax=ax, color="red", linewidth=1.2)
    rivers_region.plot(ax=ax, color="black", linewidth=0.5)

    ax.legend(
        handles=[
            mpatches.Patch(color="lightgray", label="Non-water"),
            mpatches.Patch(color="orange", label="Intermittent"),
            mpatches.Patch(color="blue", label="Perennial"),
            Line2D([0], [0], color="black", linewidth=0.5, label="HydroRIVERS"),
        ],
        loc="upper right",
        title="Legend",
    )
    ax.set_xlim(config.lon_min, config.lon_max)
    ax.set_ylim(config.lat_min, config.lat_max)
    ax.set_title(
        f"Aggregated Flow Status with HydroRIVERS, DIS_AV_CMS >= {config.discharge_threshold_cms}, "
        f"factor={config.factor}"
    )
    ax.set_xlabel("Longitude")
    ax.set_ylabel("Latitude")
    return fig, ax


def plot_classified_rivers(flow_status, extent, region_gdf, rivers_classified, config, ax=None):
    if ax is None:
        fig, ax = plt.subplots(figsize=(9, 9))
    else:
        fig = ax.figure

    cmap, norm = flow_status_colormap()
    ax.imshow(flow_status, extent=extent, origin="upper", cmap=cmap, norm=norm, alpha=0.45)
    region_gdf.boundary.plot(ax=ax, color="black", linewidth=1.0)

    rivers_classified[rivers_classified["segment_status"].isna()].plot(
        ax=ax, color="gray", linewidth=0.6, alpha=0.6
    )
    rivers_classified[rivers_classified["segment_status"] == 1].plot(
        ax=ax, color="red", linewidth=1.5
    )
    rivers_classified[rivers_classified["segment_status"] == 2].plot(
        ax=ax, color="green", linewidth=1.5
    )
    rivers_classified[rivers_classified["connected_perennial"]].plot(
        ax=ax, color="green", linewidth=2.8
    )

    ax.legend(
        handles=[
            Line2D([0], [0], color="red", linewidth=1.5, label="Intermittent river segment"),
            Line2D([0], [0], color="green", linewidth=1.5, label="Perennial river segment"),
            Line2D([0], [0], color="green", linewidth=2.8, label="Perennial by connected path"),
            Line2D([0], [0], color="gray", linewidth=0.8, label="No water grid in buffer"),
        ],
        loc="upper right",
        title="River segment status",
    )
    ax.set_xlim(config.lon_min, config.lon_max)
    ax.set_ylim(config.lat_min, config.lat_max)
    ax.set_aspect("equal")
    ax.set_title(f"River Segment Flow Status by {config.buffer_m / 1000:.0f} km Buffer + Perennial Connectivity")
    ax.set_xlabel("Longitude")
    ax.set_ylabel("Latitude")
    return fig, ax
```

```python
def load_hydrorivers(gdb_path, layer=None):
    layers = fiona.listlayers(gdb_path)
    selected_layer = layers[0] if layer is None else layer
    rivers = gpd.read_file(gdb_path, layer=selected_layer)
    return rivers, selected_layer


def make_region_gdf(bounds):
    lon_min, lat_min, lon_max, lat_max = bounds
    return gpd.GeoDataFrame(geometry=[box(lon_min, lat_min, lon_max, lat_max)], crs="EPSG:4326")


def prepare_rivers_for_region(rivers, bounds, discharge_threshold_cms):
    if rivers.crs is None:
        raise ValueError("HydroRIVERS data has no CRS.")

    rivers_plot = rivers.to_crs("EPSG:4326") if rivers.crs.to_string() != "EPSG:4326" else rivers.copy()
    large_rivers = rivers_plot[rivers_plot["DIS_AV_CMS"] >= discharge_threshold_cms].copy()
    region_gdf = make_region_gdf(bounds)
    rivers_region = gpd.clip(large_rivers, region_gdf)

    return rivers_region, region_gdf


def get_main_linestring(geom):
    if geom is None or geom.is_empty:
        return None
    if geom.geom_type == "LineString":
        return geom
    if geom.geom_type == "MultiLineString":
        lines = list(geom.geoms)
        return max(lines, key=lambda line: line.length) if lines else None
    return None


def endpoint_neighborhood_mask(point_xy, shape, transform, radius_cells):
    height, width = shape
    x, y = point_xy

    try:
        r, c = rowcol(transform, x, y)
    except Exception:
        return np.zeros(shape, dtype=bool)

    mask = np.zeros(shape, dtype=bool)
    r0 = max(0, r - radius_cells)
    r1 = min(height, r + radius_cells + 1)
    c0 = max(0, c - radius_cells)
    c1 = min(width, c + radius_cells + 1)
    mask[r0:r1, c0:c1] = True
    return mask


def has_perennial_connection_from_start_to_end(
    geom,
    flow_status,
    buffer_mask,
    agg_transform,
    endpoint_radius_cells,
    min_component_cells=2,
):
    line = get_main_linestring(geom)
    if line is None:
        return False

    coords = list(line.coords)
    if len(coords) < 2:
        return False

    perennial_mask = (flow_status == 2) & buffer_mask
    if perennial_mask.sum() < min_component_cells:
        return False

    labeled, num_features = label(perennial_mask, structure=np.ones((3, 3), dtype=int))
    if num_features == 0:
        return False

    start_mask = endpoint_neighborhood_mask(coords[0], flow_status.shape, agg_transform, endpoint_radius_cells)
    end_mask = endpoint_neighborhood_mask(coords[-1], flow_status.shape, agg_transform, endpoint_radius_cells)

    for component_id in range(1, num_features + 1):
        component = labeled == component_id
        if component.sum() < min_component_cells:
            continue
        if np.any(component & start_mask) and np.any(component & end_mask):
            return True

    return False
```

```python
def classify_river_segments_by_buffer_with_connection(
    rivers_region,
    flow_status,
    agg_transform,
    buffer_m,
    endpoint_radius_cells,
    majority_perennial_threshold,
    fallback_metric_crs="EPSG:32650",
):
    """Classify each river segment using local flow-status cells and perennial connectivity."""
    rivers_classified = rivers_region.copy().reset_index(drop=True)

    if len(rivers_classified) == 0:
        print("No river segments in this region.")
        return rivers_classified

    if rivers_classified.crs is None:
        raise ValueError("rivers_region has no CRS.")

    if rivers_classified.crs.to_string() != "EPSG:4326":
        rivers_classified = rivers_classified.to_crs("EPSG:4326")

    try:
        metric_crs = rivers_classified.estimate_utm_crs()
    except Exception:
        metric_crs = fallback_metric_crs

    buffers_lonlat = gpd.GeoSeries(
        rivers_classified.to_crs(metric_crs).geometry.buffer(buffer_m),
        crs=metric_crs,
    ).to_crs("EPSG:4326")

    rows = []
    height, width = flow_status.shape

    for geom, buffer_geom in zip(rivers_classified.geometry, buffers_lonlat):
        buffer_mask = geometry_mask(
            [buffer_geom],
            out_shape=(height, width),
            transform=agg_transform,
            invert=True,
            all_touched=True,
        )

        intermittent_count = int(np.sum((flow_status == 1) & buffer_mask))
        perennial_count = int(np.sum((flow_status == 2) & buffer_mask))
        water_count = intermittent_count + perennial_count

        if water_count == 0:
            rows.append({
                "intermittent_count": intermittent_count,
                "perennial_count": perennial_count,
                "water_count": water_count,
                "perennial_ratio": np.nan,
                "connected_perennial": False,
                "segment_status": np.nan,
                "segment_label": "No water grid",
            })
            continue

        perennial_ratio = perennial_count / water_count
        connected_perennial = has_perennial_connection_from_start_to_end(
            geom=geom,
            flow_status=flow_status,
            buffer_mask=buffer_mask,
            agg_transform=agg_transform,
            endpoint_radius_cells=endpoint_radius_cells,
            min_component_cells=2,
        )

        if connected_perennial:
            segment_status = 2
            segment_label = "Perennial_connected"
        elif perennial_ratio >= majority_perennial_threshold:
            segment_status = 2
            segment_label = "Perennial_ratio"
        else:
            segment_status = 1
            segment_label = "Intermittent_ratio"

        rows.append({
            "intermittent_count": intermittent_count,
            "perennial_count": perennial_count,
            "water_count": water_count,
            "perennial_ratio": perennial_ratio,
            "connected_perennial": connected_perennial,
            "segment_status": segment_status,
            "segment_label": segment_label,
        })

    attrs = pd.DataFrame(rows)
    return rivers_classified.join(attrs)
```

```python
def run_jrc_workflow(config):
    season, season_transform, source_crs = read_seasonality_window(config.season_path, config.bounds)
    water_agg, water_ratio, valid_count = aggregate_water_binary(
        season,
        factor=config.factor,
        threshold=config.water_ratio_threshold,
    )
    flow_status, perennial_ratio, water_count, perennial_count = aggregate_perennial_status(
        season,
        water_agg,
        factor=config.factor,
        perennial_threshold=config.perennial_month_threshold,
        ratio_threshold=config.perennial_ratio_threshold,
    )
    agg_transform = aggregated_transform(season_transform, config.factor)
    extent_agg = raster_extent(agg_transform, flow_status.shape)

    return {
        "season": season,
        "season_transform": season_transform,
        "source_crs": source_crs,
        "water_agg": water_agg,
        "water_ratio": water_ratio,
        "valid_count": valid_count,
        "flow_status": flow_status,
        "perennial_ratio": perennial_ratio,
        "water_count": water_count,
        "perennial_count": perennial_count,
        "agg_transform": agg_transform,
        "extent_agg": extent_agg,
    }


def run_river_workflow(config, flow_status, agg_transform):
    rivers, layer = load_hydrorivers(config.hydrorivers_gdb_path)
    rivers_region, region_gdf = prepare_rivers_for_region(
        rivers,
        bounds=config.bounds,
        discharge_threshold_cms=config.discharge_threshold_cms,
    )
    rivers_classified = classify_river_segments_by_buffer_with_connection(
        rivers_region=rivers_region,
        flow_status=flow_status,
        agg_transform=agg_transform,
        buffer_m=config.buffer_m,
        endpoint_radius_cells=config.endpoint_radius_cells,
        majority_perennial_threshold=config.majority_perennial_threshold,
        fallback_metric_crs=config.fallback_metric_crs,
    )

    return {
        "rivers": rivers,
        "hydrorivers_layer": layer,
        "rivers_region": rivers_region,
        "region_gdf": region_gdf,
        "rivers_classified": rivers_classified,
    }
```
