# Flow Status Mosaic

Merge saved flow-status GeoTIFF tiles and export a preview image.


```python
from pathlib import Path

import matplotlib.patches as mpatches
import matplotlib.pyplot as plt
import numpy as np
import rasterio
from matplotlib.colors import BoundaryNorm, ListedColormap
from rasterio.merge import merge
```

```python
tile_dir = Path("../../Output/FlowStatusTiles")
output_dir = Path("../../Output/FlowStatusMosaic")
output_dir.mkdir(parents=True, exist_ok=True)

tile_paths = [
    tile_dir / "flow_status_110_30_115_35_factor33.tif",
    tile_dir / "flow_status_115_30_120_35_factor33.tif",
    tile_dir / "flow_status_110_35_115_40_factor33.tif",
    tile_dir / "flow_status_115_35_120_40_factor33.tif",
]

missing = [path for path in tile_paths if not path.exists()]
if missing:
    raise FileNotFoundError(f"Missing tile files: {missing}")

mosaic_tif_path = output_dir / "flow_status_110_30_120_40_factor33.tif"
mosaic_png_path = output_dir / "flow_status_110_30_120_40_factor33.png"
```

```python
datasets = [rasterio.open(path) for path in tile_paths]
try:
    mosaic, mosaic_transform = merge(datasets, nodata=255)
    profile = datasets[0].profile.copy()
    profile.update(
        height=mosaic.shape[1],
        width=mosaic.shape[2],
        transform=mosaic_transform,
        nodata=255,
        compress="lzw",
    )

    with rasterio.open(mosaic_tif_path, "w", **profile) as dst:
        dst.write(mosaic)
        dst.update_tags(
            1,
            value_0="non-water",
            value_1="intermittent",
            value_2="perennial",
            value_255="nodata",
        )
finally:
    for dataset in datasets:
        dataset.close()

flow_status = mosaic[0]
print("Saved mosaic GeoTIFF:", mosaic_tif_path)
print("Mosaic shape:", flow_status.shape)
print("Values:", np.unique(flow_status))
print("Non-water count:", np.sum(flow_status == 0))
print("Intermittent count:", np.sum(flow_status == 1))
print("Perennial count:", np.sum(flow_status == 2))
```

```python
west = mosaic_transform.c
north = mosaic_transform.f
east = west + mosaic_transform.a * flow_status.shape[1]
south = north + mosaic_transform.e * flow_status.shape[0]
extent = [west, east, south, north]

masked_flow_status = np.ma.masked_equal(flow_status, 255)
flow_cmap = ListedColormap(["lightgray", "orange", "blue"])
flow_norm = BoundaryNorm([-0.5, 0.5, 1.5, 2.5], flow_cmap.N)

fig, ax = plt.subplots(figsize=(10, 10))
ax.imshow(
    masked_flow_status,
    extent=extent,
    origin="upper",
    cmap=flow_cmap,
    norm=flow_norm,
)
ax.legend(
    handles=[
        mpatches.Patch(color="lightgray", label="Non-water"),
        mpatches.Patch(color="orange", label="Intermittent"),
        mpatches.Patch(color="blue", label="Perennial"),
    ],
    loc="upper right",
    title="Flow status",
)
ax.set_xlim(110, 120)
ax.set_ylim(30, 40)
ax.set_aspect("equal")
ax.set_title("Flow Status Mosaic, 110-120E / 30-40N")
ax.set_xlabel("Longitude")
ax.set_ylabel("Latitude")
fig.tight_layout()
fig.savefig(mosaic_png_path, dpi=300)
plt.show()

print("Saved preview PNG:", mosaic_png_path)
```
